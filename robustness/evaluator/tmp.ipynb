```python
import pickle
import os
import math
import random

DATASETS = [
    "../../evaluation_results/robustness/helpdesk/last_event_attack_all/robustness_results.pkl",
    "../../evaluation_results/camargo/helpdesk/last_event_attack/robustness_results.pkl",
]
FRACTION = 0.005
```

```python
for src_path in DATASETS:
    with open(src_path, "rb") as f:
        data = pickle.load(f)

    n_keep = max(1, math.ceil(len(data) * FRACTION))
    keys = random.sample(list(data.keys()), n_keep)
    small = {k: data[k] for k in keys}

    # Output path: same directory + "_small" suffix, same filename
    src_dir, fname = os.path.split(src_path)
    dst_dir = src_dir + "_small"
    os.makedirs(dst_dir, exist_ok=True)
    dst_path = os.path.join(dst_dir, fname)

    with open(dst_path, "wb") as f:
        pickle.dump(small, f)

    print(f"Saved {len(small)} / {len(data)} observations to {dst_path}")
```
