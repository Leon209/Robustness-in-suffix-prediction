# Saliency Map - Attack Success Rate

Floating bar plot showing attack success rate ranges per attack type.

```python
import matplotlib.pyplot as plt
import numpy as np

x_axis = ["Attack (A1)", "Attack (A2)", "Attack (A3)", "Attack (A4)"]
y_axis = [(0.0, 0.204), (0.0, 0.25), (0.1, 0.28), (0.01, 0.38)]

x_pos = np.arange(len(x_axis))
bottom = [y[0] for y in y_axis]
height = [y[1] - y[0] for y in y_axis]

fig, ax = plt.subplots(figsize=(8, 5))
bars = ax.bar(x_pos, height, bottom=bottom, width=0.6)
ax.set_xticks(x_pos)
ax.set_xticklabels(x_axis)
ax.set_xlabel("Attack type")
ax.set_ylabel("Attack success rate")
ax.set_ylim(0, 0.45)
plt.tight_layout()
plt.show()
```
