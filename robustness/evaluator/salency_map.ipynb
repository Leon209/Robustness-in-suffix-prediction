# Saliency Map - Attack Success Rate

Floating bar plot showing attack success rate ranges per attack type.

```python
import matplotlib.pyplot as plt
import numpy as np

x_axis = ["Attack (A1)", "Attack (A2)", "Attack (A3)", "Attack (A4)"]
y_axis = [(0.0, 0.1, 0.204), (0.0, 0.12, 0.25), (0.1, 0.19, 0.28), (0.01, 0.26, 0.38)]  # (min, mean, max)

y_min = np.array([y[0] for y in y_axis])
y_mean = np.array([y[1] for y in y_axis])
y_max = np.array([y[2] for y in y_axis])
x_pos = np.arange(len(x_axis))
# Asymmetric error: below mean, above mean
yerr = np.array([y_mean - y_min, y_max - y_mean])

fig, ax = plt.subplots(figsize=(8, 5))
ax.errorbar(x_pos, y_mean, yerr=yerr, fmt="none", ecolor="black", capsize=5, capthick=1.5, linewidth=1.5)
ax.scatter(x_pos, y_mean, color="steelblue", s=60, zorder=3)
ax.set_xticks(x_pos)
ax.set_xticklabels(x_axis)
ax.set_xlabel("Attack type")
ax.set_ylabel("Attack success rate")
ax.set_xlim(-0.5, len(x_axis) - 0.5)
ax.set_ylim(0, 0.45)
ax.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```
